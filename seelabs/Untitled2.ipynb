```python
import plotly.express as px 
import pandas as pd 
```

```python
df = pd.read_csv('datasets/gapminder.csv')
df.head()
```

```python
px.line(df.query("continent=='Oceania'"), x='year', y='lifeExp', color='country', title='Oceania Life Exp').show()
```

```python
px.bar(df.query("year==2007 and continent=='Europe'"), x='country', y='gdpPercap', color='gdpPercap', title='Europe 2007').show()
```

```python
px.scatter(df.query("year==2007"), x='gdpPercap', y='lifeExp', color='continent', log_x=True,size_max=60, size='pop').show()
```
