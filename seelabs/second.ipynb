```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
import scipy.stats as stats
```

```python
iris = pd.read_csv('datasets/iris.csv')
iris.head()
```

```python
iris['sepal_r'] = iris['sepal_length']/iris['sepal_width']
iris['petal_r'] = iris['petal_length'] / iris['petal_width']
```

```python
corr = iris.corr(numeric_only=True)
sns.heatmap(corr, annot=True, fmt='.2f')
plt.title('Corr M')
plt.show()
```

```python
g = sns.PairGrid(iris, height=2.5)
g.map_diag(sns.histplot, kde=True)
g.map_upper(sns.scatterplot)
g.map_lower(sns.regplot)
g.add_legend()
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.regplot(data=iris, order=3, x='sepal_length', y='petal_length')
plt.title('Reg Plot')
plt.show()
```

```python
c_val , p_val = stats.pearsonr(iris['sepal_length'], iris['petal_length'])
round(c_val, 6)
```
