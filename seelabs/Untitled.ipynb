```python
import calendar
import pandas as pd 
import seaborn as sns 
import matplotlib.pyplot as plt 
```

```python
df = pd.read_csv('datasets/AirPassengers.csv')
```

```python
df['Month'] = pd.to_datetime(df['Month'])
df.set_index('Month', inplace=True)
```

```python
df['12_ma'] = df['Passengers'].rolling(window=12).mean() 
df.head(15)
```

```python
plt.figure(figsize=(10,5))
plt.plot(df['Passengers'], label='Actual')
plt.plot(df['12_ma'], color='red', label='12 Month Avg')
plt.title('Trend Analysis')
plt.legend()
plt.show()
```

```python
df['Month_name'] = df.index.month_name()
```

```python
months = list(calendar.month_name)[1:]
plt.figure(figsize=(10,5))
sns.boxplot(data=df, order=months, x='Month_name', y='Passengers',palette='viridis')
plt.xticks(rotation=45)
plt.title('Box Plot')
plt.show()
```
